## ⚙️ Introduction to Azure Data Factory (ADF)

## What is Azure Data Factory (ADF)?
* Azure Data Factory is a cloud-native, serverless **PaaS (Platform as a Service)** data integration and orchestration tool.

## 🔄 ETL vs. ELT (Data Processing Approaches)

When moving data from one place to another, data engineers use two primary approaches:

### 1. ETL (Extract, Transform, Load) - *Traditional Approach*
* **Extract**: Pull raw data from the source systems (e.g., SQL databases, APIs, CSV files).
* **Transform**: Clean, filter, and restructure the data *before* loading it into the destination warehouse. This transformation usually happens on an intermediate processing engine (like a staging server).
* **Load**: Save the cleaned data into the destination (e.g., Data Warehouse or Data Lake).
* *Best used when*: You need to clean or mask sensitive data before it touches storage for security and compliance.

### 2. ELT (Extract, Load, Transform) - *Modern Big Data Approach*
* **Extract**: Pull raw data from the source systems.
* **Load**: **Load the raw, unwashed data directly into the destination storage first** (like an ADLS Gen2 Data Lake).
* **Transform**: Perform the cleaning and transformations *inside* the powerful cloud destination engine (such as Azure Synapse Analytics, Databricks, or Snowflake) using cloud compute power.
* *Best used when*: Working with massive (Big Data) datasets because cloud storage is cheap and cloud engines process transformations much faster.

## 🛤️ Source ➔ ETL / ELT Pipeline Flow

Every data engineering pipeline follows this fundamental three-step journey:

1. **Source (Where the data comes from)**:
   * The origin point of your data. 
   * *Examples*: Relational databases (SQL Server, MySQL), SaaS applications, REST APIs, on-premises file servers, or cloud storage.
2. **ETL / ELT Processing (The Engine / Movement Layer)**:
   * The orchestration layer (managed by **Azure Data Factory**) that reads from the source, applies rules/transformations, and orchestrates the movement.
3. **Destination / Sink (Where the data goes)**:
   * The final target location where downstream users (data analysts, data scientists, reporting tools) can query the data.
   * *Examples*: Azure Data Lake Storage (ADLS Gen2), Azure Synapse Analytics, Azure SQL Database, or Power BI.

## Azure Data Lake vs. Blob Storage (Within a Storage Account)
When setting up storage in Azure, you choose between standard object storage and an analytics-optimized data lake:
* **Blob Storage (Default)**: Optimized for general object storage (storing images, documents, and backups with a flat namespace using slash-separated paths like `folder/file.csv`). Containers are the basic storage containers here.
* **Azure Data Lake Storage (ADLS Gen2)**: Created by checking **"Enable hierarchical namespace"** during storage account creation. It organizes data into true directory and folder structures, making it lightning-fast and cost-effective for big data analytics engines like Apache Spark and Databricks.

* **Configure Git later**: Check this box if you want to build pipelines directly in the Azure portal first without setting up GitHub or Azure DevOps immediately.

* **Validation & Creation**: Azure performs an automated configuration check. Once it passes, click **Create** to provision your Azure Data Factory instance.

### What is a Linked Service?
* **Definition**: A Linked Service in Azure Data Factory is essentially a **connection string** that defines the credentials and connection details needed to connect ADF to an external data store or compute service.
* **Simple Explanation**: Think of a Linked Service as the **"plug or cable"** that connects your Data Factory to your database, data lake, or API (e.g., connecting securely to an Azure SQL Database or an ADLS Gen2 Storage Account using usernames, passwords, or connection strings).

### What is a Dataset?
* **Definition**: A Dataset represents a specific view or structure of data within a linked data store. It points to the actual tables, files, folders, or containers using the connection provided by the Linked Service.
* **Simple Explanation**: While the Linked Service defines *where* you are connecting, the Dataset defines *what data structure* you are looking at (e.g., a specific SQL table, a folder of CSV files, or a Parquet file).

### 🔄 Data Migration Flow: Source to Destination

When migrating data using Azure Data Factory, data flows through a structured connection pipeline:

$$\text{Source (Where data lives)} \xrightarrow{\text{Linked Service}} \text{Azure Data Factory} \xrightarrow{\text{Linked Service}} \text{Destination (Where data goes)}$$

1. **Source**: The origin database, storage, or SaaS app containing the raw data.
2. **Linked Service (Source Connection)**: Authenticates and connects ADF to the source system.
3. **Azure Data Factory (Pipeline / Copy Activity)**: The orchestration engine that reads the source data and moves it.
4. **Linked Service (Destination Connection)**: Authenticates and connects ADF to the target system.
5. **Destination (Sink)**: The target data lake, warehouse, or database where the migrated data is stored.

## 🔌 Steps to Create a New Linked Service in Azure Data Factory

### Step-by-Step Navigation in ADF Studio:
1. **Navigate to the Manage Hub**: In the left-hand vertical navigation sidebar of the Data Factory Studio, click on the **Manage (⚙️)** icon.
2. **Select Linked Services**: Under the *Connections* section in the left menu of the Manage hub, click on **Linked services**.

4. **Choose Your Data Store / Compute Type**: 
   * A gallery of supported connectors will appear (e.g., *Azure SQL Database*, *Azure Data Lake Storage Gen2*, *Amazon S3*, *REST*, etc.). 
   * Search for or select the service you want to connect to (e.g., **Azure SQL Database** or **Azure Data Lake Storage Gen2**) and click **Continue**.

   * **Test Connection**: Always click the **Test connection** button to verify that ADF can successfully reach and authenticate with your source or destination before saving.
6. **Create / Save**: Click **Create** to save your Linked Service, making it ready to use across your pipelines and datasets.

### What is the Copy Activity?
* **Definition**: The **Copy Activity** is the core, serverless data movement engine inside Azure Data Factory used to copy data securely and at scale between a source and a destination (sink).
* **Simple Explanation**: Think of the Copy Activity as the **"heavy-duty moving truck"** of ADF. While pipelines handle the scheduling and workflow logic, the Copy Activity does the actual heavy lifting of reading data from your source, handling format conversions, and writing it into your destination.

## ✏️ The Author Hub & Core Components

In the Azure Data Factory Studio, clicking the **Author (✏️)** icon on the left navigation bar opens your development and design workspace. This is where you build, configure, and orchestrate all your data integration artifacts.

### 🔑 Key Components Found in the Author Hub

1. **Pipelines**:
   * **What it is**: A logical grouping or workflow of activities that together perform a task. 
   * **Simple Explanation**: Think of a Pipeline as a container or a workflow chart. For example, a pipeline can contain a *Copy Activity* to ingest data, followed by a *Data Flow* to clean it.

2. **Change Data Capture (CDC)**:
   * **What it is**: A feature in ADF that tracks and captures only the data that has changed (inserts, updates, deletes) in a source database since the last run.
   * **Why it matters**: Instead of moving entire multi-gigabyte tables every single day (full load), CDC allows you to perform lightning-fast, incremental loads, saving time and compute costs.

3. **Datasets**:
   * **What it is**: Named views of data that point to or reference the data within your Linked Services.
   * **Simple Explanation**: While a Linked Service connects ADF to your database, a Dataset defines the exact structure (like a specific SQL table or folder of CSV files) that your pipeline will read from or write to.

4. **Data Flows (Mapping Data Flows)**:
   * **What it is**: Code-free, visual data transformation workflows executed on scale-out Apache Spark compute clusters managed entirely by Azure.
   * **Why it matters**: Allows data engineers to build complex data transformations (joins, aggregations, filtering, pivoting) visually without needing to write or manage PySpark/Scala code.

5. **Power Query**:
   * **What it is**: An interactive, code-free data wrangling experience integrated directly into ADF.
   * **Why it matters**: Uses the familiar Power Query Online interface (popular in Excel and Power BI) to prototype and execute data cleaning steps, which ADF automatically translates behind the scenes into scalable Spark code.

## 🛠️ Pipeline Activities

When building a pipeline in the ADF Authoring canvas, **Activities** represent the individual processing steps or actions you want to execute. They are grouped into several logical categories:

1. **Move and Transform**:
   * *What it is*: Activities used to ingest, copy, and transform data at scale.
   * *Key Examples*: **Copy Activity** (for moving data) and **Mapping Data Flows** (for code-free transformations).

2. **Synapse**:
   * *What it is*: Activities designed to interact directly with Azure Synapse Analytics workspaces.
   * *Key Use Case*: Triggering Synapse SQL scripts or Spark notebooks directly from your ADF pipeline.

3. **Azure Data Explorer**:
   * *What it is*: Activities used to run control or data commands against Azure Data Explorer (Kusto) clusters for real-time log and telemetry analytics.

4. **Azure Function**:
   * *What it is*: Allows you to invoke custom serverless code written in an Azure Function (Python, C#, JavaScript) as a step inside your data pipeline.

5. **Batch Service**:
   * *What it is*: Runs custom parallel and high-performance computing (HPC) batch jobs on managed Azure compute pools.

6. **Databricks**:
   * *What it is*: Integrates Apache Spark workloads by allowing you to execute Notebooks, JARs, or Python scripts directly within your ADF pipeline.

7. **General**:
   * *What it is*: Core utility and control-flow activities used to manage pipeline execution.
   * *Key Examples*: *Wait*, *Set Variable*, *Web activity* (to call REST APIs), and *Append variable*.

8. **HDInsight**:
   * *What it is*: Activities used to execute legacy Hadoop ecosystem jobs (Hive, Pig, MapReduce, or Spark) on Azure HDInsight clusters.

9. **Iteration & Conditionals**:
   * *What it is*: Control flow logic used to build smart, dynamic workflows.
   * *Key Examples*: **ForEach** (for looping through lists of files or tables), **Until** (looping until a condition is met), and **If Condition** (branching logic based on success or failure).

10. **Machine Learning**:
    * *What it is*: Activities used to execute trained machine learning models or scoring experiments hosted in Azure Machine Learning.

11. **Power Query**:
    * *What it is*: An activity that executes visual, code-free data wrangling scripts built using the Power Query interface inside the pipeline.

## 🔗 Azure Data Factory: Activity Dependency Paths

### 1. What are Activity Dependency Paths?
* **Definition**: Activity dependency paths are visual connection lines in Azure Data Factory that define the execution order and condition under which a subsequent (downstream) activity runs based on the outcome of the preceding (upstream) activity.
* **Simple Explanation**: Think of dependency paths as **"traffic signals and directional arrows"** connecting your pipeline boxes together. By default, drawing an arrow connects them *On success*, but you can change the color and rule so that different paths trigger depending on whether the previous step succeeded, failed, was skipped, or simply finished running.

### 🔑 The Four Dependency Types (Color-Coded Arrows)

1. **On Success (Green Arrow - Default)**:
   * *When it runs*: The downstream activity executes **only if** the upstream activity completes successfully without any errors.
   * *Use Case*: Running a data copy job only after a "Get Metadata" check confirms the source file exists.

2. **On Failure (Red Arrow)**:
   * *When it runs*: The downstream activity executes **only if** the upstream activity fails or throws an error.
   * *Use Case*: Triggering an error-logging script, a notification email, or an alert activity if your main pipeline copy job fails.

3. **On Skip (Gray / Light Blue Arrow)**:
   * *When it runs*: The downstream activity executes **only if** the upstream activity was skipped (usually due to a conditional branch evaluation where that path wasn't chosen).
   * *Use Case*: Handling alternative fallback flows when a specific step is bypassed.

4. **On Completion (Blue Arrow)**:
   * *When it runs*: The downstream activity executes **regardless** of whether the upstream activity succeeded, failed, or was skipped.
   * *Use Case*: Running a cleanup task, closing a connection, or logging audit metrics at the end of a process no matter how the previous step turned out.

### ⚙️ How to Configure Dependency Paths in ADF
1. **Connect Activities**: Drag a connection icon (the small square box with an arrow handle on the edge of an activity) from an upstream activity and drop it onto the downstream activity.
2. **Change Dependency Condition**: 
   * Click on the connecting arrow line between the two activities.
   * A small settings prompt or properties menu will appear allowing you to check or unchecking the boxes for **Success**, **Failure**, **Skipped**, or **Completed** to instantly change the rule and arrow color.

## 🔍 Get Metadata Activity

### 1. What is the Get Metadata Activity?
* **Definition**: The **Get Metadata** activity is a control-flow activity in Azure Data Factory used to retrieve metadata (information *about* data, rather than the data itself) from any supported data store or file.
* **Simple Explanation**: Think of Get Metadata as a **"scout or detective"** in your pipeline. Before your pipeline attempts to copy or transform a file or table, this activity checks things like:
  * Does the file or table actually exist?
  * How big is the file (in bytes)?
  * How many columns or rows does it have?
  * What is the structure (schema) or last modified timestamp?

### ⚙️ How to Configure Get Metadata in ADF
1. **Add Activity**: In the Author tab, expand the **General** category in the activities panel and drag **Get metadata** onto your pipeline canvas.
2. **Dataset Tab**: 
   * Select an existing dataset (or create a new one pointing to your file or table) that you want to inspect.
3. **Field List Tab**: 
   * Click **+ New** to select the specific metadata properties you want to extract. Common options include:
    * **Child items**: Returns a list of all files and subfolders located inside a specified container or directory.
    * **Exists**: Returns a boolean (`true` or `false`) indicating whether the specified file, table, or folder actually exists.
    * **Item name**: Returns the name of the file or folder being inspected.
    * **Item type**: Specifies whether the inspected item is a File or a Folder (Container).
    * **Last modified**: Returns a timestamp showing the exact date and time the file or folder was last updated.
    * **Add dynamic content**: An expression builder tool used to write code referencing outputs and variables from previous activities.
4. **Using the Output**: 
   * You can reference the output of a Get Metadata activity in subsequent activities using dynamic content expressions (e.g., `@activity('GetMetadata1').output.exists` or `@activity('GetMetadata1').output.childItems`).

## 🔀 Azure Data Factory: If Condition Activity

### 1. What is the If Condition Activity?
* **Definition**: The **If Condition** activity is a control-flow activity in Azure Data Factory that evaluates a boolean expression (yielding `true` or `false`) to determine which set of downstream activities should be executed.
* **Simple Explanation**: Think of the If Condition activity as a **"fork in the road"** or a traffic cop in your pipeline. It looks at a condition (like checking if a file exists or if a row count is greater than zero) and directs the pipeline down one of two paths:
  * **True Path**: Executes if the evaluated expression is true.
  * **False Path**: Executes if the evaluated expression is false.

### ⚙️ How to Configure If Condition in ADF
1. **Add Activity**: In the Author tab, expand the **Iteration & conditionals** category in the activities panel and drag **If Condition** onto your pipeline canvas.
2. **Settings Tab (The Expression)**: 
   * Click into the **Expression** box and click **Add dynamic content** to write the logical condition (e.g., `@activity('GetMetadata1').output.exists` or `@equals(activity('GetMetadata1').output.itemType, 'Folder')`).
3. **Activities Tab (True / False Blocks)**: 
   * Inside the If Condition activity properties, click on the pencil icon for the **True** activities block to drop in the steps that should run if the expression evaluates to true.
   * Do the same for the **False** activities block to build out the alternative execution path.

## 🔁 Azure Data Factory: ForEach Activity

### 1. What is the ForEach Activity?
* **Definition**: The **ForEach** activity is a control-flow activity in Azure Data Factory used to execute a set of activities repeatedly in a loop over a collection of items (such as a list of file names, table names, or folder paths).
* **Simple Explanation**: Think of the ForEach activity as a **"conveyor belt or automated loop"** in your pipeline. Instead of manually writing a copy activity for every single file or table, you feed it a list (like the output of a Get Metadata `childItems` list), and it loops through them one by one (or concurrently) to perform the same action on each item.

### ⚙️ How to Configure ForEach in ADF
1. **Add Activity**: In the Author tab, expand the **Iteration & conditionals** category in the activities panel and drag **For each** onto your pipeline canvas.
2. **Settings Tab**:
   * **Batch count**: Controls how many items are processed simultaneously (leave unchecked for default sequential processing, or set a number for parallel execution up to 50).
   * **Items**: Click into the box, open **Add dynamic content**, and provide the array or collection you want to loop through (e.g., pointing to the output of a Get Metadata activity: `@activity('GetMetadata1').output.childItems`).
3. **Activities Tab (Inside the Loop)**:
   * Click on the pencil icon inside the ForEach activity box to open its internal canvas.
   * Drag and drop the activities you want repeated for each item (such as a **Copy Activity**).
   * Inside those nested activities, use the loop item variable—`@item().name`—to dynamically reference the current file or table being processed in that iteration.

## 🎯 Hands-On Lab: Filtering and Copying Specific Files Using ADF Control Flow

### Overview of the Scenario
When a container or folder contains multiple files, but you only want to extract a specific subset (e.g., files starting with `"Fact"`), you combine **Get Metadata**, **ForEach**, **If Condition**, and a **Parameterized Copy Activity** to automate the selective migration.

---

### Step-by-Step Implementation Flow

#### 1. Step 1: Inspect Folder Contents (Get Metadata Activity)
* **Pipeline Setup**: Created a new pipeline and dragged the **Get Metadata** activity onto the canvas.
* **Dataset Configuration**: Pointed it to the source container/folder.
* **Field List**: Selected **Child items** to retrieve a complete array/list of all files and folders present in the directory.

#### 2. Step 2: Loop Through All Files (ForEach Activity)
* **Add Activity**: Dragged the **For each** activity and connected it to the Get Metadata activity.
* **Settings Configuration**:
  * Checked **Sequential** (to process files one by one).
  * In the **Items** box, used dynamic content to pull the file list from the previous step:
    ```text
    @activity('Get Metadata1').output.childItems
    ```

#### 3. Step 3: Filter Files by Name (If Condition Activity)
* **Inside ForEach**: Clicked inside the ForEach activity and dragged the **If Condition** activity (`iffilematches`).
* **Expression Configuration**: Used dynamic content to check if each iterated file name starts with `"Fact"`:
  ```text
  @startswith(item().name, 'Fact')
  ```

#### 4. Parameterized Copy Activity (True Activities)
* Inside the **True** activities block, dragged and dropped the **Copy data** activity.
* **Source Configuration**:
  * Set the source to a parameterized source pointing to the folder where all files are present.
  * Created a dataset parameter named `pfile` and set the file path using dynamic content: `pfile`.
  * In the Copy Activity's source settings for parameter `pfile`, added the dynamic value:
    ```text
    @item().name
    ```
* **Sink Configuration**:
  * Pointed the sink to the `reporting` container.
  * Created a dataset parameter named `pfile2` and set the file path in the dataset to `@dataset().pfile2`.
  * In the Copy Activity's sink settings for parameter `pfile2`, added the value:
    ```text
    @item().name
    ```

## ⚙️ Azure Data Factory: Data Flow Activity

### 1. What is the Data Flow Activity?
* **Definition**: The **Data Flow** activity is a pipeline-level control flow action used to execute a previously designed Mapping Data Flow within an ADF pipeline.
* **Simple Explanation**: Think of a Mapping Data Flow as the *recipe or blueprint* (the transformation logic), while the **Data Flow Activity** is the *cook or execution button* placed inside your pipeline that actually runs that recipe using an Azure-managed Spark cluster.

### ⚙️ How to Add and Configure a Data Flow Activity
1. **Create Pipeline**: Open your pipeline canvas in the Author tab.
2. **Add Activity**: Expand the **Move & transform** category in the activities panel and drag the **Data flow** activity onto your pipeline canvas.
3. **Configure Settings**: 
   * Click on the activity, go to the **Settings** tab, and select your target Data Flow from the dropdown.
   * Choose your Integration Runtime compute properties and configure any parameters if your data flow requires them.
4. **Execute**: Connect it to upstream activities (like a *Get Metadata* or *Wait* activity) or run a **Debug** session to test the Spark-powered transformation live.
