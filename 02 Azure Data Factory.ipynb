## ⚙️ Introduction to Azure Data Factory (ADF)

## What is Azure Data Factory (ADF)?
* Azure Data Factory is a cloud-native, serverless **PaaS (Platform as a Service)** data integration and orchestration tool.

## 🔄 ETL vs. ELT (Data Processing Approaches)

When moving data from one place to another, data engineers use two primary approaches:

### 1. ETL (Extract, Transform, Load) - *Traditional Approach*
* **Extract**: Pull raw data from the source systems (e.g., SQL databases, APIs, CSV files).
* **Transform**: Clean, filter, and restructure the data *before* loading it into the destination warehouse. This transformation usually happens on an intermediate processing engine (like a staging server).
* **Load**: Save the cleaned data into the destination (e.g., Data Warehouse or Data Lake).
* *Best used when*: You need to clean or mask sensitive data before it touches storage for security and compliance.

### 2. ELT (Extract, Load, Transform) - *Modern Big Data Approach*
* **Extract**: Pull raw data from the source systems.
* **Load**: **Load the raw, unwashed data directly into the destination storage first** (like an ADLS Gen2 Data Lake).
* **Transform**: Perform the cleaning and transformations *inside* the powerful cloud destination engine (such as Azure Synapse Analytics, Databricks, or Snowflake) using cloud compute power.
* *Best used when*: Working with massive (Big Data) datasets because cloud storage is cheap and cloud engines process transformations much faster.

## 🛤️ Source ➔ ETL / ELT Pipeline Flow

Every data engineering pipeline follows this fundamental three-step journey:

1. **Source (Where the data comes from)**:
   * The origin point of your data. 
   * *Examples*: Relational databases (SQL Server, MySQL), SaaS applications, REST APIs, on-premises file servers, or cloud storage.
2. **ETL / ELT Processing (The Engine / Movement Layer)**:
   * The orchestration layer (managed by **Azure Data Factory**) that reads from the source, applies rules/transformations, and orchestrates the movement.
3. **Destination / Sink (Where the data goes)**:
   * The final target location where downstream users (data analysts, data scientists, reporting tools) can query the data.
   * *Examples*: Azure Data Lake Storage (ADLS Gen2), Azure Synapse Analytics, Azure SQL Database, or Power BI.

## Azure Data Lake vs. Blob Storage (Within a Storage Account)
When setting up storage in Azure, you choose between standard object storage and an analytics-optimized data lake:
* **Blob Storage (Default)**: Optimized for general object storage (storing images, documents, and backups with a flat namespace using slash-separated paths like `folder/file.csv`). Containers are the basic storage containers here.
* **Azure Data Lake Storage (ADLS Gen2)**: Created by checking **"Enable hierarchical namespace"** during storage account creation. It organizes data into true directory and folder structures, making it lightning-fast and cost-effective for big data analytics engines like Apache Spark and Databricks.

## 🏭 Steps to Create an Azure Data Factory

### Step-by-Step Navigation:
1. **Search for Data Factories**: Log in to the [Azure Portal](https://portal.azure.com), go to the top search bar, type `Data factories`, and select it from the services list.
2. **Initiate Creation**: Click on the **+ Create** button.

### 🔍 Configuring the Data Factory Tabs

#### 1. Basics Tab
The Basics tab defines the core identity, subscription, region, and version of your Data Factory.
* **Project Details**:
  * **Subscription**: Select your active Azure billing subscription (e.g., *Free Trial* or *Pay-As-You-Go*).
  * **Resource Group**: Select your designated resource group (e.g., `RG-AzureFundamentals`).
* **Instance Details**:
  * **Name**: Enter a globally unique name for your data factory (e.g., `adf-dataengineering-prod`).
  * **Region**: Choose the Azure region where your ADF instance will be hosted (e.g., `Central India` or `Canada Central`).
  * **Version**: 
    * **V2**: The current, recommended production-grade standard that supports enterprise-scale data integration, SSIS migration, and mapping data flows.

#### 2. Git Configuration Tab
Allows you to connect your Data Factory to a source control repository for collaborative development and version tracking.
* **Configure Git later**: Check this box if you want to build pipelines directly in the Azure portal first without setting up GitHub or Azure DevOps immediately.
* **Repository type**: Choose between *GitHub* or *Azure DevOps* to manage and version-control your pipeline JSON code and scripts.

#### 3. Networking Tab
Controls secure access and communication boundaries for your data factory.
* **Public endpoint**: Allows connectivity via the public internet, secured using Microsoft Entra ID and role-based access control (RBAC).
* **Managed Virtual Network (VNet)**: Allows ADF to securely connect to data stores via private endpoints without exposing traffic to the public internet, keeping enterprise data isolated and secure.

#### 4. Advanced Tab
* Configures encryption settings, allowing you to use either **Microsoft-managed keys** or **Customer-managed keys** via Azure Key Vault to encrypt metadata and credentials at rest.

#### 5. Tags Tab
* **Definition**: Name/value pairs (metadata) assigned to resources for billing, tracking, and organization (e.g., `Environment: Development`, `Project: DataEngineering`).

#### 6. Review + Create Tab
* **Validation & Creation**: Azure performs an automated configuration check. Once it passes, click **Create** to provision your Azure Data Factory instance.

### What is a Linked Service?
* **Definition**: A Linked Service in Azure Data Factory is essentially a **connection string** that defines the credentials and connection details needed to connect ADF to an external data store or compute service.
* **Simple Explanation**: Think of a Linked Service as the **"plug or cable"** that connects your Data Factory to your database, data lake, or API (e.g., connecting securely to an Azure SQL Database or an ADLS Gen2 Storage Account using usernames, passwords, or connection strings).

### What is a Dataset?
* **Definition**: A Dataset represents a specific view or structure of data within a linked data store. It points to the actual tables, files, folders, or containers using the connection provided by the Linked Service.
* **Simple Explanation**: While the Linked Service defines *where* you are connecting, the Dataset defines *what data structure* you are looking at (e.g., a specific SQL table, a folder of CSV files, or a Parquet file).

### 🔄 Data Migration Flow: Source to Destination

When migrating data using Azure Data Factory, data flows through a structured connection pipeline:

$$\text{Source (Where data lives)} \xrightarrow{\text{Linked Service}} \text{Azure Data Factory} \xrightarrow{\text{Linked Service}} \text{Destination (Where data goes)}$$

1. **Source**: The origin database, storage, or SaaS app containing the raw data.
2. **Linked Service (Source Connection)**: Authenticates and connects ADF to the source system.
3. **Azure Data Factory (Pipeline / Copy Activity)**: The orchestration engine that reads the source data and moves it.
4. **Linked Service (Destination Connection)**: Authenticates and connects ADF to the target system.
5. **Destination (Sink)**: The target data lake, warehouse, or database where the migrated data is stored.

## 🔌 Steps to Create a New Linked Service in Azure Data Factory

### Step-by-Step Navigation in ADF Studio:
1. **Navigate to the Manage Hub**: In the left-hand vertical navigation sidebar of the Data Factory Studio, click on the **Manage (⚙️)** icon.
2. **Select Linked Services**: Under the *Connections* section in the left menu of the Manage hub, click on **Linked services**.
3. **Initiate New Connection**: Click on the **+ New** button at the top of the Linked Services pane.
4. **Choose Your Data Store / Compute Type**: 
   * A gallery of supported connectors will appear (e.g., *Azure SQL Database*, *Azure Data Lake Storage Gen2*, *Amazon S3*, *REST*, etc.). 
   * Search for or select the service you want to connect to (e.g., **Azure SQL Database** or **Azure Data Lake Storage Gen2**) and click **Continue**.
5. **Configure Linked Service Properties**:
   * **Name**: Enter a clear, descriptive name for your connection (e.g., `LS_AzureSQL_Target` or `LS_ADLSGen2_DataLake`).
   * **Integration runtime**: Select your integration runtime (default is typically `AutoResolveIntegrationRuntime`, which runs fully managed in the cloud).
   * **Authentication method**: Choose how ADF authenticates with the service (e.g., *Account Key*, *Connection String*, *Service Principal*, or *Managed Identity*).
   * **Test Connection**: Always click the **Test connection** button to verify that ADF can successfully reach and authenticate with your source or destination before saving.
6. **Create / Save**: Click **Create** to save your Linked Service, making it ready to use across your pipelines and datasets.

## ✏️ The Author Hub & Core Components

In the Azure Data Factory Studio, clicking the **Author (✏️)** icon on the left navigation bar opens your development and design workspace. This is where you build, configure, and orchestrate all your data integration artifacts.

### 🔑 Key Components Found in the Author Hub

1. **Pipelines**:
   * **What it is**: A logical grouping or workflow of activities that together perform a task. 
   * **Simple Explanation**: Think of a Pipeline as a container or a workflow chart. For example, a pipeline can contain a *Copy Activity* to ingest data, followed by a *Data Flow* to clean it.

2. **Change Data Capture (CDC)**:
   * **What it is**: A feature in ADF that tracks and captures only the data that has changed (inserts, updates, deletes) in a source database since the last run.
   * **Why it matters**: Instead of moving entire multi-gigabyte tables every single day (full load), CDC allows you to perform lightning-fast, incremental loads, saving time and compute costs.

3. **Datasets**:
   * **What it is**: Named views of data that point to or reference the data within your Linked Services.
   * **Simple Explanation**: While a Linked Service connects ADF to your database, a Dataset defines the exact structure (like a specific SQL table or folder of CSV files) that your pipeline will read from or write to.

4. **Data Flows (Mapping Data Flows)**:
   * **What it is**: Code-free, visual data transformation workflows executed on scale-out Apache Spark compute clusters managed entirely by Azure.
   * **Why it matters**: Allows data engineers to build complex data transformations (joins, aggregations, filtering, pivoting) visually without needing to write or manage PySpark/Scala code.

5. **Power Query**:
   * **What it is**: An interactive, code-free data wrangling experience integrated directly into ADF.
   * **Why it matters**: Uses the familiar Power Query Online interface (popular in Excel and Power BI) to prototype and execute data cleaning steps, which ADF automatically translates behind the scenes into scalable Spark code.

##  Monitor
The **Monitor** tab is the operational dashboard in ADF Studio. It is used to track the health, status, and performance of your data workflows and compute infrastructure. 

Key monitoring components include:

*   **Pipeline runs:** A historical and real-time log of every time a pipeline executes. It displays the status (Succeeded, Failed, In Progress), execution duration, and detailed error messages.
*   **Trigger runs:** A view that tracks executions initiated by automated triggers (such as a scheduled time or a file arrival event), ensuring your scheduled jobs are firing off correctly.
*   **Change data capture (CDC):** A monitoring interface for CDC resources. CDC is a process that tracks and extracts only the incremental data that has changed (inserts, updates, deletes) in a source system, rather than copying the entire dataset every time.
*   **Integration runtimes (IR):** A dashboard to check the health and status (e.g., Running, Offline) of your Integration Runtimes. The IR is the underlying compute infrastructure that actually executes or dispatches your data movement and transformation activities.
*   **Data flow debug:** A view to manage active debug sessions. These are temporary Spark clusters spun up so developers can preview data and interactively test their transformations before publishing them.
*   **Alerts & metrics:** 
    *   **Metrics:** Quantifiable data points collected by Azure (like the number of successful pipeline runs or IR CPU usage).
    *   **Alerts:** Automated notifications (like an email or SMS) configured to trigger when a specific event occurs, such as a pipeline failure or a metric crossing a predefined threshold.

## Manage Hub

The **Manage** tab (⚙️) is the central administration and configuration hub for your Azure Data Factory. It is where you configure global settings, connections, security, and version control.

### 1. General
*   **Factory settings:** The core properties and billing information for your ADF instance, including subscription details, resource group, and default configurations.
*   **Connector upgrade:** A section to manage, review, and upgrade the underlying drivers or connectors ADF uses to interact with external data sources, ensuring you have the latest features and security patches.

### 2. Connections
*   **Linked service:** The connection strings and authentication credentials required for ADF to connect to external data stores (like databases or storage accounts) and compute resources.
*   **Integration runtime (IR):** The compute infrastructure that actually executes or dispatches your data movement (Copy) and transformation activities.
*   **Microsoft Purview:** An integration setting that connects your ADF to Microsoft Purview, enabling automatic data lineage tracking and governance across your pipelines.
*   **ADF in Microsoft Fabric:** Settings to integrate and manage your Data Factory workloads within the broader Microsoft Fabric analytics ecosystem.

### 3. Source control
*   **Git configuration:** The interface to connect your ADF workspace to a version control repository (GitHub or Azure DevOps). This enables collaborative development, branch management, and CI/CD (Continuous Integration/Continuous Deployment).
*   **ARM template:** A tool to export your entire Data Factory configuration as an Azure Resource Manager (ARM) JSON template. This is heavily used to deploy your ADF setup from a Development environment to Test or Production environments.

### 4. Author
*   **Triggers:** The schedulers or event-listeners (like time schedules, tumbling windows, or storage events) that automate when your pipelines execute.
*   **Global parameters:** Constants or variables defined at the factory level that can be consumed by any pipeline across the entire ADF instance (e.g., storing an environment name like "Dev" or "Prod").
*   **Data flow libraries:** Reusable custom functions, logic, or code libraries that you can define once and apply across multiple Mapping Data Flows to standardize transformations.

### 5. Security
*   **Credentials:** Securely managed authentication identities (such as Managed Identities or Service Principals) used by ADF to access external secure resources, like an Azure Key Vault.
*   **Customer managed key:** An encryption setting that allows you to secure your ADF metadata using your own encryption keys stored in Azure Key Vault, rather than relying on Microsoft's default managed keys.
*   **Outbound rules:** Network security configurations used when your ADF is running inside a Managed Virtual Network (VNet). These rules define and restrict which external endpoints or IP addresses your Data Factory is permitted to communicate with.

## 🔗 Activity Dependency Paths

### 1. What are Activity Dependency Paths?
* **Definition**: Activity dependency paths are visual connection lines in Azure Data Factory that define the execution order and condition under which a subsequent (downstream) activity runs based on the outcome of the preceding (upstream) activity.
* **Simple Explanation**: Think of dependency paths as **"traffic signals and directional arrows"** connecting your pipeline boxes together. By default, drawing an arrow connects them *On success*, but you can change the color and rule so that different paths trigger depending on whether the previous step succeeded, failed, was skipped, or simply finished running.

### 🔑 The Four Dependency Types (Color-Coded Arrows)

1. **On Success (Green Arrow - Default)**:
   * *When it runs*: The downstream activity executes **only if** the upstream activity completes successfully without any errors.
   * *Use Case*: Running a data copy job only after a "Get Metadata" check confirms the source file exists.

2. **On Failure (Red Arrow)**:
   * *When it runs*: The downstream activity executes **only if** the upstream activity fails or throws an error.
   * *Use Case*: Triggering an error-logging script, a notification email, or an alert activity if your main pipeline copy job fails.

3. **On Skip (Gray / Light Blue Arrow)**:
   * *When it runs*: The downstream activity executes **only if** the upstream activity was skipped (usually due to a conditional branch evaluation where that path wasn't chosen).
   * *Use Case*: Handling alternative fallback flows when a specific step is bypassed.

4. **On Completion (Blue Arrow)**:
   * *When it runs*: The downstream activity executes **regardless** of whether the upstream activity succeeded, failed, or was skipped.
   * *Use Case*: Running a cleanup task, closing a connection, or logging audit metrics at the end of a process no matter how the previous step turned out.

### ⚙️ How to Configure Dependency Paths in ADF
1. **Connect Activities**: Drag a connection icon (the small square box with an arrow handle on the edge of an activity) from an upstream activity and drop it onto the downstream activity.
2. **Change Dependency Condition**: 
   * Click on the connecting arrow line between the two activities.
   * A small settings prompt or properties menu will appear allowing you to check or unchecking the boxes for **Success**, **Failure**, **Skipped**, or **Completed** to instantly change the rule and arrow color.

### What is a Trigger?
* **Definition**: A trigger in Azure Data Factory determines when and how a pipeline is executed. 
* **Simple Explanation**: While clicking the **Debug** button runs your pipeline manually for testing, a **Trigger** automates your pipeline so it runs on a schedule, when a new file lands in storage, or in response to custom events without manual intervention.

### What is a Storage Events Trigger?
* **Definition**: A **Storage Events Trigger** (often called a Blob Events Trigger) is an event-driven trigger in Azure Data Factory that automatically kicks off a pipeline the exact moment a file is created, updated, or deleted in an Azure Storage account container.
* **Simple Explanation**: Instead of running your pipeline on a rigid time schedule (like "run every day at 2 PM"), a storage events trigger acts like a **doorbell**. The moment a new data file lands in your storage container, Azure Event Grid notifies ADF, and your pipeline fires instantly.

## ⚙️ Integration Runtime (IR) Setup

### What is an Integration Runtime?
The Integration Runtime (IR) is the compute infrastructure used by Azure Data Factory to provide data integration capabilities across different network environments. 

### ➕ Creating a New Integration Runtime
To create a new IR, click on **+ New** and select the type of Integration Runtime based on your compute requirements:
* **Azure** (Fully managed cloud compute)
* **Self-Hosted** (For on-premises or private network data)
* **Azure-SSIS** (To execute SSIS packages)
* **Airflow** (Managed Airflow environment)

#### Network Environment Options:
* Azure
* Self-Hosted
* **External Resources**: Linked Self-Hosted

## 🛠️ Self-Hosted Integration Runtime Setup

When you choose to create a **Self-Hosted** Integration Runtime (to connect to on-premises data), follow these configuration steps:

### 1. Settings
* **Name**: Provide a unique name for your Integration Runtime.
* **Self-contained interactive authoring**: Choose to **Enable** or **Disable** this feature based on your security and authoring needs.
* Click **Create**.

### 2. Setup Options
Once created, you will be presented with two ways to install the IR on your local machine or VM:

* **Option 1: Express Setup** (Quick, automated installation directly to your machine)
* **Option 2: Manual Setup** (Preferred for controlled environments)

### 3. Manual Setup Steps
If you choose the Manual Setup, follow this two-step process:

* **Step 1: Download and Install**
  * Click the link to download and install the Integration Runtime. This will open a new window.
  * Choose the specific download version you want and run the installer on your local system.
* **Step 2: Authentication Keys**
  * ADF will generate authentication keys (**Key 1** and **Key 2**). You will need one of these to link your local machine to ADF.

### 4. Local System Configuration
Once the software is installed on your local machine:
1. Search your system for **Microsoft Integration Runtime Configuration Manager** and open it.
2. **Register Integration Runtime (Self-Hosted)**: Copy **Key 1** (or Key 2) from the Azure portal and paste it into the authentication box here.
3. **(Optional) HTTP Proxy**: Configure proxy settings if your corporate network requires it to reach the internet.
4. **Diagnostic Tool**: Use the built-in diagnostic tool to test connections and troubleshoot any network issues.

✅ **Result:** Once successfully registered and running, **now we can use on-prem data** securely in our Azure Data Factory pipelines!

## What is Data Orchestration?
Data orchestration is the automated coordination, scheduling, and management of complex data pipelines. Instead of triggering individual pipelines (like API ingestion, On-Prem ingestion, or SQL ingestion) manually, we wrap them into a single **Parent (Master) Pipeline**. This allows us to define execution sequences, pass parameters dynamically, manage dependencies, and set up automated schedules or alerts.

## What is a Logic App and Why Use It?
When running automated data pipelines, you need a mechanism to notify the on-call team immediately if a critical failure occurs. Azure Data Factory (ADF) does not have a native "Send Email" activity. 
Instead, we use an **Azure Logic App**—a cloud-based web application capable of receiving HTTP requests (like a REST API endpoint) and executing subsequent actions, such as dispatching an email. ADF will essentially make an API call to the Logic App whenever a pipeline fails.
