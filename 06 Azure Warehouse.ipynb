# 06. Azure Data Warehouse

## 1. Data Warehousing Fundamentals

### What is a Data Warehouse?
* A **central repository** for all of an organization's data, acting as the single source of truth for analytics and reporting (Power BI, Tableau, Qlik).
* **Why not build reports directly on source systems?** Pointing heavy analytical queries directly at transactional data sources degrades their performance.

### On-Premises vs. Cloud Data Warehouse
* **On-Premises DW:** Hosted on physical servers or Virtual Machines (VMs), such as running SQL Server / SSMS on a VM. Works for small datasets, but hard to scale.
* **Why move to a Cloud Data Warehouse?**
  * **Storage & Cost:** Much cheaper to store rapidly growing data.
  * **High Availability:** Cloud providers handle replication (LRS, GRS, GZRS) and guarantee uptime (e.g., 99.95%+).
  * **Performance:** Uses **distributed computing/querying** to process data across multiple machines.
  * **PaaS (Platform as a Service):** No manual upgrades or hardware maintenance.
  * **Security:** Built-in firewalls, virtual networks, and access controls.
  * **Easy Integration:** Simple connectors to pull from and push to multiple cloud/on-prem platforms.

**Popular Cloud Solutions:**
* **Azure:** Azure Synapse Analytics *(end-to-end solution, formerly Azure SQL Data Warehouse)*
* **AWS:** Amazon Redshift
* **GCP:** Google BigQuery
* **Multi-Cloud:** Snowflake (DW), Databricks (Lakehouse)

### Database (OLTP) vs. Data Warehouse (OLAP)

| Feature | Database (e.g., Azure SQL DB) | Data Warehouse (e.g., Azure Synapse) |
| :--- | :--- | :--- |
| **Primary Purpose** | Transactions (OLTP) | Analytics & Reporting (OLAP) |
| **Optimized For** | Frequent **writes** (inserts, updates, deletes in milliseconds) | Heavy **reads** (querying 1–10 years of historical data) |
| **Impact** | Running big reports disrupts live operations | Isolated from live apps; handles large aggregations easily |

### What is Cloud Data Warehousing?
* The end-to-end process of integrating data from multiple sources (**SQL DBs, APIs, on-prem files, ERP apps like Dynamics 365, Dataverse**) into a central Data Warehouse using **ETL / ELT**.
* **Batch Loading:** Instead of constantly querying the source DB, data is pulled in scheduled batches (e.g., hourly or a few times a day) and loaded incrementally so the source system isn't overloaded.

### ETL & Data Warehouse Layers (vs. Medallion Architecture)
**ETL** = **E**xtract → **T**ransform → **L**oad

1. **Staging Layer *(Bronze)***
   * Stores raw, **as-is** replicas of the source data.
   * **Transient:** Temporary storage (cleared after processing).
   * **Persistent:** Permanent storage (keeps historical raw data).
2. **Transform Layer *(Silver)***
   * Applies cleaning and business logic (removing/modifying/adding columns).
   * Optional in traditional ETL, but saved as the **Silver** layer in modern Medallion Architecture.
3. **Core Layer *(Gold)***
   * The actual **Data Warehouse**, where transformed data is structured into a **Data Model** (Facts & Dimensions, or One Big Table) for fast reporting.
4. **Data Marts**
   * Smaller subsets of the main Data Warehouse filtered for specific business units (e.g., a **Finance** Data Mart or **HR** Data Mart) so teams only see relevant data.
