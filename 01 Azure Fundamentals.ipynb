
## Cloud: 

Center location to store the data. A vast global network of physical servers in secure data centers worldwide. Instead of storing data or running applications locally or on-premises, resources (compute, storage, network) are accessed remotely over the internet.

## Cloud Computing: 

The on-demand delivery of IT resources (compute power, storage, databases) over the internet with pay-as-you-go pricing, enabling dynamic scaling without maintaining physical data centers.

Cloud computing is the delivery of computing services—such as servers, storage, databases, networking, and analytics—over the internet. Instead of buying and maintaining your own hardware, you use a provider’s resources when you need them and pay according to your usage.

### Key characteristics

• **On-demand self-service** Create or configure resources yourself, without waiting for someone to provision hardware. Create a virtual machine or database through the Azure portal or CLI.

• **Broad network access** Services are available over a network using common devices and tools. Access Azure Storage from a laptop, application, or data pipeline.

• **Resource pooling** The provider’s infrastructure serves multiple customers, while keeping their data and resources isolated. Many customers use Azure datacenters without sharing access to one another’s data.

• **Rapid elasticity** Resources can scale up or down as demand changes. Add compute capacity during a large data-processing job, then reduce it afterward.

• **Measured service** Usage is monitored so costs can be measured and billed. Pay for storage capacity or compute time used.

### Cloud deployment models

• **Public cloud** A cloud provider, such as Microsoft Azure, owns and operates the infrastructure, which is used by multiple customers with logical separation.

• **Private cloud** Cloud infrastructure dedicated to one organization, hosted in its own datacenter or by a provider.

•**Hybrid cloud** A combination of private infrastructure and public cloud, connected so data or applications can work across both.

• **Multi-cloud** Using services from more than one public cloud provider, such as Azure and AWS.

*Hybrid cloud and multi-cloud are different concepts, and an organization can use both.*

## Cloud Service Models
• **IaaS (Infrastructure as a Service)**: Microsoft manages physical hardware and data centers; customer manages OS, runtime, and apps (e.g., Azure VM).

• **PaaS (Platform as a Service)**: Microsoft manages hardware, OS, patching, and scaling; customer focuses on code and data (e.g., Azure Data Factory, Azure Databricks, Azure Synapse). Primary model for Data Engineers.

• **SaaS (Software as a Service)**: Vendor manages the full stack end-to-end; customer simply uses the software (e.g., Power BI, Microsoft 365).

## Physical Infrastructure & Availability

• **Geography**: A defined area of the world that contains one or more Azure regions. Azure geographies help support data residency and compliance requirements. Check the specific service’s terms and configuration to confirm where data is stored or processed.

• **Datacenter**: An individual physical facility containing server racks, power, and cooling.

• **Availability Zone (AZ)**: Physically isolated location within a region made up of one or more datacenters with independent power, cooling, and networking.

• **Data Availability Zone**: Zone-Redundant architectures ensuring High Availability (HA) and Fault Tolerance. Data in ADLS Gen2 (via ZRS) is synchronously replicated across three zones, achieving RPO = 0 if a single datacenter fails.

• **Region Pair**: Two regions within the same geography separated by 300+ miles for Disaster Recovery (DR) and sequential platform updates.

• **Edge location / Point of Presence (PoP)**: A site closer to users that helps services deliver content or route requests with lower latency. Azure Front Door and Azure CDN use Microsoft’s global edge network.

## Azure Storage Redundancy Options
• **LRS (Locally Redundant Storage)**: 3 copies in 1 datacenter. Cheapest; best for dev/test and temporary staging data.

• **ZRS (Zone-Redundant Storage)**: 3 copies across 3 availability zones in the same region. High availability against datacenter outages with zero data loss. Industry standard for ADLS Gen2.

• **GRS (Geo-Redundant Storage)**: 6 copies total (3 in primary region via LRS + 3 in paired region via LRS). Disaster recovery against regional outages.

• **GZRS (Geo-Zone-Redundant Storage)**: 6 copies total (3 across zones in primary via ZRS + 3 in paired region via LRS).

• **Production Rule**: Use ZRS for standard Data Lake storage to avoid cross-region network costs; use GRS/GZRS when enterprise DR mandates cross-region replication.
