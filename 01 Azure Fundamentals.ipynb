
# 01. Azure Fundamentals & Core Concepts

## Cloud: 

Center location to store the data. A vast global network of physical servers in secure data centers worldwide. Instead of storing data or running applications locally or on-premises, resources (compute, storage, network) are accessed remotely over the internet.

## Cloud Computing: 

The on-demand delivery of IT resources (compute power, storage, databases) over the internet with pay-as-you-go pricing, enabling dynamic scaling without maintaining physical data centers.

Cloud computing is the delivery of computing services—such as servers, storage, databases, networking, and analytics—over the internet. Instead of buying and maintaining your own hardware, you use a provider’s resources when you need them and pay according to your usage.

## 🔑 Key Characteristics of Cloud Computing
1. **On-demand self-service**: Users can provision computing resources (like virtual machines or databases) dynamically without requiring human intervention from the cloud provider.
2. **Broad network access**: Services are available over a network and accessed through standard mechanisms using heterogeneous client platforms (laptops, mobile devices, etc.).
3. **Resource pooling**: The provider’s computing resources are pooled to serve multiple consumers using a multi-tenant model, with different physical and virtual resources dynamically assigned and reassigned according to demand.
4. **Rapid elasticity**: Capabilities can be elastically provisioned and released to scale outward and inward rapidly based on demand.
5. **Measured service**: Cloud systems automatically control and optimize resource use by leveraging a metering capability at some level of abstraction appropriate to the type of service (e.g., storage, compute, bandwidth).

## 🌐 Cloud Deployment Models

* **Public Cloud**: Owned and operated by a third-party cloud service provider (e.g., Microsoft Azure, AWS), delivering computing resources like servers and storage over the internet.
* **Private Cloud**: Cloud infrastructure exclusively used by a single business or organization. It can be physically located at the organization's on-site data center or hosted by a third-party service provider.
* **Hybrid Cloud**: Combines public and private clouds, bound together by technology that allows data and applications to be shared between them.
* **Multi-Cloud**: Using multiple public cloud providers (e.g., combining Microsoft Azure and AWS) to avoid vendor lock-in and optimize workloads.

## 🛠️ Cloud Service Models

* **IaaS (Infrastructure as Service)**: 
  * *Definition*: Provides basic computing infrastructure—virtual machines (VMs), servers, storage, and networking.
  * *Management*: The cloud provider manages the physical hardware, data centers, and physical networking; you manage the OS, middleware, runtimes, and applications.
  * *Example*: Azure Virtual Machines.

* **PaaS (Platform as a Service)**: 
  * *Definition*: Provides a managed platform and environment allowing developers to build, test, and deploy applications without worrying about underlying infrastructure. **Primary model for Data Engineers**.
  * *Management*: The provider manages the hardware, OS, patching, and scaling; you focus purely on your application code and data configurations.
  * *Example*: Azure Data Factory, Azure Databricks, Azure SQL Database.

* **SaaS (Software as a Service)**: 
  * *Definition*: Delivers software applications over the internet, on-demand, typically via a subscription model.
  * *Management*: The vendor manages everything (servers, storage, networking, runtime, data, and application code); you simply use the software.
  * *Example*: Power BI, Microsoft 365.

## 🌍 Azure Physical Infrastructure & Availability

* **Region**: A set of datacenters deployed within a latency-defined perimeter and connected through a dedicated regional low-latency network.
* **Availability Zone (AZ)**: Physically separate locations within an Azure region made up of one or more datacenters equipped with independent power, cooling, and networking to ensure high availability.
* **Region Pair**: Each Azure region is paired with another region within the same geography (at least 300 miles away) to provide reliable replication and disaster recovery (DR).
* **Geography**: A discrete market, typically spanning two or more regions, that preserves data residency and compliance boundaries.

## 🗄️ Azure Storage Redundancy Options

* **LRS (Locally Redundant Storage)**: Replicates your data 3 times synchronously within a single datacenter in the primary region. Lowest cost, but least resilient against facility-wide outages.
* **ZRS (Zone-Redundant Storage)**: Replicates your data synchronously across 3 Azure availability zones in the primary region. Recommended for high availability (HA) for Data Lake storage (ADLS Gen2).
* **GRS (Geo-Redundant Storage)**: Copies your data synchronously 3 times in the primary region (via LRS), and then copies your data asynchronously 3 times to a secondary paired region.
* **GZRS (Geo-Zone-Redundant Storage)**: Combines ZRS (across 3 zones in the primary region) with GRS (asynchronous replication to the paired region). Ideal for maximum durability and enterprise disaster recovery.

## 🗂️ Organizing Azure Resources

* **Management Groups**: Hierarchical containers placed above subscriptions that allow organizations to centrally manage access, compliance, and policies across multiple subscriptions.
* **Subscriptions**: Logical billing and administrative units linked to an account used to provision, track, and pay for Azure services.
* **Resource Groups**: Fundamental logical containers that hold related resources for an Azure solution, serving as a boundary for lifecycle control, access management (RBAC), and billing.
* **Resources**: Individual, manageable cloud services and instances (such as a Virtual Machine or Storage Account) deployed within a resource group.

## 📦 Azure Resources and Resource Groups

### 1. What is an Azure Resource?
* **Definition**: An individual, manageable service or instance hosted in the cloud (such as an Azure SQL Database, an Azure Storage Account, or a Virtual Machine).
* **Simple Explanation**: Think of a single resource as an individual item or appliance in your house—like a refrigerator or a TV. 

### 2. What is a Resource Group?
* **Definition**: A logical container that holds related resources for an Azure solution. 
* **Simple Explanation**: Think of a resource group as a **cardboard moving box** or a folder on your computer where you group related items together. 
* **Key Purpose**: 
  * **Lifecycle Management**: If you delete a resource group, *all* the resources inside it are deleted together (making it easy to clean up test environments).
  * **Security & Billing**: You can apply access control (IAM) and track costs at the resource group level.

## 📁 Creating a Resource Group in Azure

### Steps via Azure Portal:
1. **Log in to the Azure Portal**: Navigate to [portal.azure.com](https://portal.azure.com) and sign in with your Azure account.
2. **Search for Resource Groups**: In the top search bar, type `Resource groups` and select it from the search results.
3. **Create a New Resource Group**: Click on the **+ Create** button at the top of the Resource Groups page.
4. **Configure Basics**:
   * **Subscription**: Select your active Azure subscription (e.g., *Free Trial* or *Pay-As-You-Go*).
   * **Resource group**: Enter a unique name for your resource group (e.g., `RG-AzureFundamentals` or `rg-dataengineering-prod`).
   * **Region**: Choose the Azure region closest to you or where you plan to deploy your resources (e.g., `Central US` or `South India`).
5. **Review and Create**: Click **Review + create**, verify your settings, and then click **Create**.

## 📊 Data Formats

* **Structured Data**: Data that adheres to a strict, predefined schema or data model (such as rows and columns in relational databases or SQL tables). It is highly organized and easy to search, query, and analyze (e.g., CSV files, relational tables).
* **Semi-Structured Data**: Data that does not conform to a rigid relational structure (tables), but contains organizational markers, tags, or metadata (such as key-value pairs or hierarchies) that make it easier to parse and process (e.g., JSON, XML, YAML).
* **Unstructured Data**: Data that lacks any predefined data model or organizational structure, making it difficult to process with traditional tools without AI or specialized processing (e.g., text documents, images, audio files, videos).

## 🗄️ Steps to Create an Azure SQL Database

### Step-by-Step Navigation:
1. **Search for SQL Databases**: Log in to the Azure Portal, go to the top search bar, type `SQL databases`, and select it from the services list.
2. **Initiate Creation**: Click on the **+ Create** button.
3. **Select Architecture / Deployment Path**: Upon starting, you will encounter two main options/approaches:
   * **Azure SQL Database Hyperscale**
   * **Migrate to Azure Cloud**

### 🔍 Understanding the Two Options

#### 1. Azure SQL Database Hyperscale
* **What it is**: A highly scalable, cloud-native service tier within Azure SQL Database designed to handle massive data growth.
* **How it works**: It decouples storage from compute using a distributed architecture. 
* **Key Benefits**: 
  * Storage scales dynamically up to 100 TB+ without being constrained by local node sizes.
  * Rapid database backups and lightning-fast scaling operations regardless of the size of the data.

#### 2. Migrate to Azure Cloud
* **What it is**: A guided deployment pathway and integration with tools (such as the **Azure Database Migration Service**) designed for moving existing databases from on-premises environments or virtual machines into the managed Azure ecosystem.
* **How it works**: It assists you in assessing compatibility, transferring schemas, and replicating data with minimal downtime.
* **Key Benefits**: 
  * Simplifies "lift-and-shift" or modernizing legacy SQL Server workloads.
  * Transition from self-managed infrastructure to a fully managed PaaS (Platform as a Service) environment.

  * 🖥️ Creating an Azure SQL Server

      * **Use both SQL and Microsoft Entra authentication**: Allows access using either Microsoft Entra identities or traditional SQL server admin credentials.
      * **Use SQL authentication**: Relies entirely on traditional database administrator usernames and passwords.

* Workload Environment
  * **Development**: Optimized for lower costs, testing, and non-production workloads.
  * **Production**: Sets robust defaults, high availability, and performance-tuned parameters for enterprise applications.

* Compute + Storage (Service Tiers)
  Clicking **Configure database** lets you choose your architectural tier and pricing model:
  * **General Purpose**: Balanced compute and storage option designed for standard, everyday workloads.
  * **Business Critical**: Designed for low-latency, high-availability transactional workloads with heavy read/write requirements.
  * **☁️ Azure SQL Database Hyperscale**: A highly scalable service tier that leverages cloud-scale architecture to decouple storage and compute, allowing storage to scale dynamically up to 100 TB+ without being constrained by local node sizes.

* Behavior when Free Offer Limit Reached
  When utilizing free monthly tier allowances (such as 100,000 vCore seconds and 32GB storage):
  * **Auto-pause the database until next month**: The database automatically pauses and becomes inaccessible once the free limit is hit, preventing any extra charges until the next calendar month.
  * **Continue using database for additional charges**: Keeps the database accessible, billing any overage usage at standard general-purpose serverless tier rates.

* Backup Storage Redundancy
  Controls how Point-In-Time Restore (PITR) and Long-Term Retention (LTR) backups are replicated for disaster recovery:
  * **Locally-redundant backup storage (LRS)**: Replicates your backups synchronously within a single physical data center.
  * **Zone-redundant backup storage (ZRS)**: Replicates backups across multiple availability zones within a region for higher availability.
  * **Geo-redundant backup storage (GRS)**: Replicates backups to a paired secondary region, enabling geo-restore capability during regional outages.
  * **Geo-Zone-redundant backup storage (GZRS)**: Combines zone redundancy with geo-replication for maximum enterprise-grade durability.

### 2. Networking Tab

* **Identity**:
  * Configures Microsoft Entra ID (formerly Azure Active Directory) authentication alongside traditional SQL authentication for secure, centralized access control.

### 4. Additional Settings Tab
This tab handles data source initialization, collation rules, and data maintenance windows.

* **Use existing data**:

  * Specifies when system updates and patching can occur (e.g., System Default, Weekday, or Weekend windows).

### 🔍 Configuring the Storage Account Tabs

#### 1. Basics Tab
The Basics tab defines the project details, instance name, performance, redundancy, and primary usage profiles.
* **Project Details**:
  * **Subscription**: Select your active Azure billing subscription.
  * **Resource Group**: Select your resource group (e.g., `RG-AzureFundamentals`).
* **Instance Details**:
  * **Storage account name**: Enter a globally unique name (e.g., `apsst`). Must be lowercase alphanumeric (3-24 characters).
  * **Region**: Choose the region where your storage account will be hosted (e.g., `(Canada) Canada Central`).
  * **Deploy to an Azure Extended Zone**: Optional setting to deploy storage resources closer to specific edge locations for ultra-low latency.
* **Primary Service (3 Options)**:
  * **Azure Blob Storage or Azure Data Lake Storage**: Optimized for object storage, unstructured data, and hierarchical namespace data lakes (ideal for big data analytics and Spark/Databricks).
  * **Azure Files**: Optimized for fully managed cloud file shares using standard SMB and NFS protocols.
  * **Other (Tables and Queues)**: Optimized for structured NoSQL table storage and message queuing components.
* **Primary Workload (6 Options)**:
  * **General purpose**: Best for custom storage configurations with non-workload-specific defaults (balanced for standard blobs, files, queues, and tables).
  * **Logs**: Optimized specifically for high-volume, append-heavy log generation and ingestion workloads.
  * **Media**: Optimized for media streaming, video encoding, and large asset distribution scenarios.
  * **High performance**: Optimized for low-latency, high-throughput transactional apps.
  * **Data warehouse / Analytics**: Optimized for large-scale analytical query processing and big data pipelines.
  * **Backup / Archive**: Optimized for cost-effective, long-term data retention and disaster recovery backups.
* **Performance**: 
  * **Standard**: Recommended for most scenarios using general-purpose v2 accounts (cost-effective for big data analytics).
  * **Premium**: Recommended for scenarios requiring very low latency and high transaction rates.
* **Redundancy**: Choose how your data is replicated for disaster recovery:
  * **Locally redundant storage (LRS)**: Replicates data 3 times synchronously within a single data center in the primary region (lowest cost).
  * **Zone-redundant storage (ZRS)**: Replicates data synchronously across multiple availability zones in the primary region.
  * **Geo-redundant storage (GRS)**: Replicates locally in the primary region, then asynchronously to a secondary paired region.
  * **Geo-Zone-redundant storage (GZRS)**: Combines zone redundancy with geo-replication for maximum durability.

#### 2. Advanced Tab (Enabling ADLS Gen2 Data Lake)

The Advanced tab is divided into specific sections depending on whether you are configuring **Azure Blob Storage** or **Azure Files**.
These settings apply exclusively to Azure Blob Storage workloads and object data management:

* **Enable hierarchical namespace**: 
  * *What it is*: The core setting that converts a standard blob storage account into an **Azure Data Lake Storage Gen2 (ADLS Gen2)**.
  * *Why it matters*: Instead of a flat namespace (where blobs are stored with long slash-separated strings like `folder/subfolder/file.csv`), it organizes objects into true directory and folder hierarchies. This is required for big data analytics engines like Apache Spark, Databricks, and Azure Synapse.
* **Enable network file system v3 (NFS v3)**: 
  * *What it is*: Allows Linux clients to mount blob containers using the standard NFS v3 protocol.
  * *Why it matters*: Ideal for high-performance computing (HPC), video rendering, and workloads requiring POSIX-compliant file system operations over blobs.
* **Enable SFTP (Secure File Transfer Protocol)**: 
  * *What it is*: Enables secure SSH File Transfer Protocol access directly to your storage account.
  * *Why it matters*: Allows external vendors or legacy systems to securely upload/download files via standard SFTP clients without needing complex custom applications.
* **Allow cross-tenant replication**: 
  * *What it is*: Permits data to be replicated from a storage account in your Microsoft Entra tenant to a storage account in a different Entra tenant (organization).
  * *Note*: *Cross-tenant replication and hierarchical namespace cannot be enabled simultaneously.*

* **Access Tiers (Blob Storage)**
Defines the default cost-to-storage optimization tier for your blobs:
  * **Hot**: Optimized for frequently accessed data and everyday usage scenarios (higher storage cost, very low access/transaction cost).
  * **Cool**: Optimized for infrequently accessed data that is stored for at least 30 days (lower storage cost, higher access cost).
  * **Cold**: Optimized for rarely accessed data and long-term backup scenarios (lowest storage cost, highest access/retrieval cost).

* **Azure Files Settings**
These settings apply exclusively to **Azure Files** (managed cloud file shares using SMB and NFS protocols):
  * **SMB Multichannel**: An optional setting (subject to account kind, performance, and replication constraints) that allows Windows clients to establish multiple simultaneous network connections to an SMB file share for higher throughput and fault tolerance.
  * **Enable Managed Identity for SMB**: Allows Microsoft Entra identities to authenticate and authorize access to Azure file shares using SMB, improving security over traditional storage account keys.
  * **Require Encryption in Transit for SMB**: Enforces encryption for all SMB traffic entering or leaving the file share, preventing man-in-the-middle network snooping.
  * **Require Encryption in Transit for NFS**: Enforces encryption for NFS file shares (subject to regional and tier support constraints).

#### 3. Networking

The Networking tab controls how clients and applications connect to your storage account, balancing accessibility and security.

* **Public Access**
Controls how your resource is accessed through public networks:
* **Public network access**:
  * **Enable**: Allows inbound and outbound access, with the option to restrict select inbound traffic using resource access configurations.
  * **Disable**: Restricts all inbound access while still allowing outbound access.
  * **Secure by perimeter (Most restricted)**: Restricts both inbound and outbound access using a network security perimeter, offering the highest level of restriction.
* **Public network access scope**:
  * **Enable from all networks**: Makes the resource accessible publicly from anywhere on the internet (increases security risk unless necessary).
  * **Enable from selected virtual networks and IP addresses**: Restricts public access so that only explicitly approved virtual networks and IP ranges can connect.

* **Private Endpoint**
  * **Definition**: Allows you to create a secure, private connection to your storage account using a private IP address from your Virtual Network (VNet). 
  * **Key Purpose**: Keeps traffic entirely within the Microsoft backbone network, avoiding exposure to the public internet entirely (recommended for enterprise production workloads).
**
* **Network Routing**
  Determines how traffic is routed as it travels from the source to its Azure endpoint:
  * **Microsoft network routing (Recommended)**: Routes traffic over Microsoft's global, high-performance private backbone network for most of the journey, ensuring optimal reliability and performance.
  * **Internet routing**: Routes traffic primarily over the public internet to reach the Azure endpoint.

#### 4. Data Protection

The Data Protection tab configures recovery, auditing, tracking, and immutability safeguards to protect your data against accidental deletion, corruption, or unauthorized modifications. These settings apply exclusively to Azure Blob Storage workloads and object data recovery:

* **Enable point-in-time restore for containers**:
  * *What it is*: Allows you to restore one or more blob containers to an earlier point in time.
  * *Prerequisites*: Requires **Versioning**, **Change feed**, and **Blob soft delete** to be enabled simultaneously.
* **Enable soft delete for blobs**:
  * *What it is*: Enables you to recover blobs and directories that were previously marked for deletion within a specific retention window.
  * *Days to retain deleted blobs*: Defines how long deleted items are kept before being permanently erased (default is typically set to `7` days).
* **Enable soft delete for containers**:
  * *What it is*: Enables recovery for entire containers that were accidentally deleted.
  * *Note*: Enabling soft delete for frequently overwritten data may increase storage costs.
  * *Days to retain deleted containers*: Retention period for deleted containers (default is `7` days).
* **Enable versioning for blobs**:
  * *What it is*: Automatically maintains previous versions of your blobs whenever a blob is modified or deleted, allowing you to revert to an older state.
* **Enable blob change feed**:
  * *What it is*: Keeps a detailed, chronological track of create, modification, and delete operations performed on blobs in your account, which is useful for streaming analytics and auditing.
* **Enable version-level immutability support**:
  * *What it is*: Allows you to set an account-level, time-based retention policy (WORM - Write Once, Read Many) that prevents data from being modified or deleted until the retention period expires.
  * *Prerequisites*: **Versioning** must be enabled for this property to be active.

* **Azure Files Data Protection Settings**
These settings apply exclusively to **Azure Files** recovery:
  * **Enable soft delete for classic file shares**:
    * *What it is*: Allows you to recover classic file shares that were previously marked for deletion.
    * *Days to retain deleted classic file shares*: Defines the retention window before a deleted file share is permanently removed.

#### 5. Security

The Security tab configures encryption standards, authentication methods, access controls, and threat protection policies for your storage account.
* **Require secure transfer for REST API operations**: 
  * *What it is*: Enforces that all incoming requests to the storage account via REST API must be made over HTTPS (TLS encryption). 
  * *Why it matters*: Highly recommended to ensure data in transit is encrypted and protected from man-in-the-middle sniffing.
* **Allow enabling anonymous access on individual containers**: 
  * *What it is*: Controls whether specific blob containers within this storage account can be configured for public anonymous read access (e.g., for static website hosting or public asset delivery).
* **Enable storage account key access**: 
  * *What it is*: Permits clients to authenticate using the master storage account access keys (Account Key 1 and Key 2). Disabling this forces the use of Microsoft Entra ID or SAS tokens for enhanced security.
* **Default to Microsoft Entra authorization in the Azure portal**: 
  * *What it is*: Sets Microsoft Entra ID (formerly Azure AD) as the default authentication method when viewing data in the Azure Portal, rather than defaulting to storage account access keys.
* **Minimum TLS version**: 
  * *What it is*: Defines the lowest version of Transport Layer Security (TLS) accepted for incoming connections (e.g., **Version 1.2**). Older, less secure protocols like TLS 1.0 or 1.1 are blocked.
* **Permitted scope for copy operations (Preview)**: 
  * *What it is*: Restricts where data can be copied to or from within storage accounts (e.g., *From any storage account* vs. restricting it strictly to accounts within the same Microsoft Entra tenant).

* **Microsoft Defender for Storage**
  * **Definition**: An Azure-native layer of security intelligence that detects potential threats against your storage accounts. 
  * **Key Purpose**: Helps prevent three major impacts on your data and workloads:
    1. **Malicious file uploads** (scanning uploaded files for malware/viruses).
    2. **Sensitive data exfiltration** (detecting unauthorized access or data theft).
    3. **Data corruption** (identifying abnormal access patterns or tampering).
  * **Enable Defender for Storage**: Check this box to activate advanced threat intelligence monitoring on your storage account.

#### 6. Encryption

The Encryption tab controls how your data at rest is secured and protected using cryptographic keys. Azure automatically encrypts all data in a storage account, but you can choose how those encryption keys are managed.
* **Microsoft-managed keys (MMK)**: 
  * *What it is*: The default and easiest option where Microsoft generates, manages, and automatically rotates the cryptographic keys protecting your data. 
  * *Security*: Provides strong security with zero operational overhead for key management.
* **Customer-managed keys (CMK)**: 
  * *What it is*: Allows you to bring your own cryptographic keys and manage them using **Azure Key Vault** or Managed HSM.
  * *Why it matters*: Gives your organization complete control over key lifecycles, rotation policies, and the ability to revoke access to your data at any time.
  * **Enable support for customer-managed keys**: 
    * *Blobs and files only*: Applies customer-managed keys exclusively to blob and file storage.
    * *All service types (blobs, files, tables, and queues)*: Applies customer-managed keys across all storage services. 
    * *Important Constraint*: **This option cannot be changed after the storage account is created.**

* **Enable Infrastructure Encryption**
  * **Definition**: Adds a second layer of encryption at rest. While standard Azure storage encryption uses a single key per block, infrastructure encryption applies an additional layer of encryption using a separate key managed by Microsoft (often adhering to FIPS 140-2 compliance standards).
  * **Key Purpose**: Provides double encryption to protect against scenarios where a single encryption algorithm or key might theoretically be compromised.
  * **Encryption**: Manage encryption keys (Microsoft-managed vs. Customer-managed keys via Azure Key Vault).
  * **Tags**: Add optional metadata tags (e.g., `Project: DataEngineering`).

#### 7. Tags
Tags are name/value pairs (metadata) assigned to resources to help you categorize them for billing, cost allocation, automation, and governance.
  * **Name**: The category key (e.g., `Environment`, `Project`, `Owner`).
  * **Separator / Value**: The specific assignment or grouping detail (e.g., `Production`, `DataEngineering`, `DataTeam`).
  * **Resource Scope**: Allows you to apply tags to the storage account itself or inherit/manage them across related child components.
  * **Delete**: Removes an assigned tag pair if no longer needed.

* **Networking & Connectivity Resources within Storage Context**
When managing security and connectivity perimeter components for your storage account, Azure integrates with several core networking services:
  * **Private Endpoint**: Secures data traffic by connecting the storage account directly to your virtual network using a private IP address, keeping all traffic off the public internet.
  * **Storage Account**: The primary cloud storage service container (Blob, ADLS Gen2, Files, Tables, Queues) holding your enterprise data.
  * **Virtual Network (VNet)**: The foundational secure private network in Azure that enables resources (like VMs, databricks, or private endpoints) to securely communicate with each other, the internet, and on-premises networks.

#### 8. Review + Create Tab
* **Validation & Creation**: Azure validates your configurations. Click **Create** to provision your data lake.

## 💾 Data Storage Services

Within an Azure Storage Account, data is organized into four primary storage services tailored for different unstructured, semi-structured, and tabular data needs:

### 1. Containers (Blob Storage / ADLS Gen2)
* **What it is**: An object storage solution optimized for storing massive amounts of unstructured data, such as text documents, images, audio, video files, and big data analytical files (Parquet, CSV, Delta).
* **Key Purpose**: Similar to a directory in a file system, a container organizes a set of blobs. When hierarchical namespace is enabled, containers support true folder and directory trees (Data Lake Gen2), making them the primary storage foundation for big data analytics engines like Apache Spark and Databricks.

### 2. Classic File Shares
* **What it is**: Fully managed cloud file shares that can be accessed via the industry-standard Server Message Block (SMB) and Network File System (NFS) protocols.
* **Key Purpose**: Allows legacy applications and virtual machines to mount file shares in the cloud just like a traditional on-premises network drive, enabling multiple VMs or users to read and write to the same files simultaneously.

### 3. Queues
* **What it is**: A messaging store designed for reliable asynchronous message queuing and storage between application components.
* **Key Purpose**: Allows decoupled cloud applications, microservices, and background workers to communicate with each other at scale by passing messages back and forth through a queue.

### 4. Tables
* **What it is**: A NoSQL key-attribute (key-value) data store designed for structured, non-relational data storage.
* **Key Purpose**: Allows applications to store massive quantities of semi-structured schema-less data (such as user profiles, logs, or device telemetry) with rapid retrieval via clustered indexes.
