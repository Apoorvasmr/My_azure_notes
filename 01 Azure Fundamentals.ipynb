
# 01. Azure Fundamentals & Core Concepts

## Cloud: 

Center location to store the data. A vast global network of physical servers in secure data centers worldwide. Instead of storing data or running applications locally or on-premises, resources (compute, storage, network) are accessed remotely over the internet.

## Cloud Computing: 

The on-demand delivery of IT resources (compute power, storage, databases) over the internet with pay-as-you-go pricing, enabling dynamic scaling without maintaining physical data centers.

Cloud computing is the delivery of computing services—such as servers, storage, databases, networking, and analytics—over the internet. Instead of buying and maintaining your own hardware, you use a provider’s resources when you need them and pay according to your usage.

## 🔑 Key Characteristics of Cloud Computing
1. **On-demand self-service**: Users can provision computing resources (like virtual machines or databases) dynamically without requiring human intervention from the cloud provider.
2. **Broad network access**: Services are available over a network and accessed through standard mechanisms using heterogeneous client platforms (laptops, mobile devices, etc.).
3. **Resource pooling**: The provider’s computing resources are pooled to serve multiple consumers using a multi-tenant model, with different physical and virtual resources dynamically assigned and reassigned according to demand.
4. **Rapid elasticity**: Capabilities can be elastically provisioned and released to scale outward and inward rapidly based on demand.
5. **Measured service**: Cloud systems automatically control and optimize resource use by leveraging a metering capability at some level of abstraction appropriate to the type of service (e.g., storage, compute, bandwidth).

## 🌐 Cloud Deployment Models

* **Public Cloud**: Owned and operated by a third-party cloud service provider (e.g., Microsoft Azure, AWS), delivering computing resources like servers and storage over the internet.
* **Private Cloud**: Cloud infrastructure exclusively used by a single business or organization. It can be physically located at the organization's on-site data center or hosted by a third-party service provider.
* **Hybrid Cloud**: Combines public and private clouds, bound together by technology that allows data and applications to be shared between them.
* **Multi-Cloud**: Using multiple public cloud providers (e.g., combining Microsoft Azure and AWS) to avoid vendor lock-in and optimize workloads.

## 🛠️ Cloud Service Models

* **IaaS (Infrastructure as Service)**: 
  * *Definition*: Provides basic computing infrastructure—virtual machines (VMs), servers, storage, and networking.
  * *Management*: The cloud provider manages the physical hardware, data centers, and physical networking; you manage the OS, middleware, runtimes, and applications.
  * *Example*: Azure Virtual Machines.

* **PaaS (Platform as a Service)**: 
  * *Definition*: Provides a managed platform and environment allowing developers to build, test, and deploy applications without worrying about underlying infrastructure. **Primary model for Data Engineers**.
  * *Management*: The provider manages the hardware, OS, patching, and scaling; you focus purely on your application code and data configurations.
  * *Example*: Azure Data Factory, Azure Databricks, Azure SQL Database.

* **SaaS (Software as a Service)**: 
  * *Definition*: Delivers software applications over the internet, on-demand, typically via a subscription model.
  * *Management*: The vendor manages everything (servers, storage, networking, runtime, data, and application code); you simply use the software.
  * *Example*: Power BI, Microsoft 365.

## 🌍 Azure Physical Infrastructure & Availability

* **Region**: A set of datacenters deployed within a latency-defined perimeter and connected through a dedicated regional low-latency network.
* **Availability Zone (AZ)**: Physically separate locations within an Azure region made up of one or more datacenters equipped with independent power, cooling, and networking to ensure high availability.
* **Region Pair**: Each Azure region is paired with another region within the same geography (at least 300 miles away) to provide reliable replication and disaster recovery (DR).
* **Geography**: A discrete market, typically spanning two or more regions, that preserves data residency and compliance boundaries.

## 🗄️ Azure Storage Redundancy Options

* **LRS (Locally Redundant Storage)**: Replicates your data 3 times synchronously within a single datacenter in the primary region. Lowest cost, but least resilient against facility-wide outages.
* **ZRS (Zone-Redundant Storage)**: Replicates your data synchronously across 3 Azure availability zones in the primary region. Recommended for high availability (HA) for Data Lake storage (ADLS Gen2).
* **GRS (Geo-Redundant Storage)**: Copies your data synchronously 3 times in the primary region (via LRS), and then copies your data asynchronously 3 times to a secondary paired region.
* **GZRS (Geo-Zone-Redundant Storage)**: Combines ZRS (across 3 zones in the primary region) with GRS (asynchronous replication to the paired region). Ideal for maximum durability and enterprise disaster recovery.

## 🗂️ Organizing Azure Resources

* **Management Groups**: Hierarchical containers placed above subscriptions that allow organizations to centrally manage access, compliance, and policies across multiple subscriptions.
* **Subscriptions**: Logical billing and administrative units linked to an account used to provision, track, and pay for Azure services.
* **Resource Groups**: Fundamental logical containers that hold related resources for an Azure solution, serving as a boundary for lifecycle control, access management (RBAC), and billing.
* **Resources**: Individual, manageable cloud services and instances (such as a Virtual Machine or Storage Account) deployed within a resource group.

## 📁 Creating a Resource Group in Azure

### Steps via Azure Portal:
1. **Log in to the Azure Portal**: Navigate to [portal.azure.com](https://portal.azure.com) and sign in with your Azure account.
2. **Search for Resource Groups**: In the top search bar, type `Resource groups` and select it from the search results.
3. **Create a New Resource Group**: Click on the **+ Create** button at the top of the Resource Groups page.
4. **Configure Basics**:
   * **Subscription**: Select your active Azure subscription (e.g., *Free Trial* or *Pay-As-You-Go*).
   * **Resource group**: Enter a unique name for your resource group (e.g., `RG-AzureFundamentals` or `rg-dataengineering-prod`).
   * **Region**: Choose the Azure region closest to you or where you plan to deploy your resources (e.g., `Central US` or `South India`).
5. **Review and Create**: Click **Review + create**, verify your settings, and then click **Create**.

## 📊 Data Formats

* **Structured Data**: Data that adheres to a strict, predefined schema or data model (such as rows and columns in relational databases or SQL tables). It is highly organized and easy to search, query, and analyze (e.g., CSV files, relational tables).
* **Semi-Structured Data**: Data that does not conform to a rigid relational structure (tables), but contains organizational markers, tags, or metadata (such as key-value pairs or hierarchies) that make it easier to parse and process (e.g., JSON, XML, YAML).
* **Unstructured Data**: Data that lacks any predefined data model or organizational structure, making it difficult to process with traditional tools without AI or specialized processing (e.g., text documents, images, audio files, videos).

## 🗄️ Steps to Create an Azure SQL Database

### Step-by-Step Navigation:
1. **Search for SQL Databases**: Log in to the Azure Portal, go to the top search bar, type `SQL databases`, and select it from the services list.
2. **Initiate Creation**: Click on the **+ Create** button.
3. **Select Architecture / Deployment Path**: Upon starting, you will encounter two main options/approaches:
   * **Azure SQL Database Hyperscale**
   * **Migrate to Azure Cloud**

---

### 🔍 Understanding the Two Options

#### 1. Azure SQL Database Hyperscale
* **What it is**: A highly scalable, cloud-native service tier within Azure SQL Database designed to handle massive data growth.
* **How it works**: It decouples storage from compute using a distributed architecture. 
* **Key Benefits**: 
  * Storage scales dynamically up to 100 TB+ without being constrained by local node sizes.
  * Rapid database backups and lightning-fast scaling operations regardless of the size of the data.

#### 2. Migrate to Azure Cloud
* **What it is**: A guided deployment pathway and integration with tools (such as the **Azure Database Migration Service**) designed for moving existing databases from on-premises environments or virtual machines into the managed Azure ecosystem.
* **How it works**: It assists you in assessing compatibility, transferring schemas, and replicating data with minimal downtime.
* **Key Benefits**: 
  * Simplifies "lift-and-shift" or modernizing legacy SQL Server workloads.
  * Transition from self-managed infrastructure to a fully managed PaaS (Platform as a Service) environment.

### 1. Basics Tab
The Basics tab defines where your database lives, its identity, and its compute/storage tier.

* **Project Details**:
  * **Subscription**: The active Azure billing subscription used to pay for the database.
  * **Resource Group**: The logical container (e.g., `RG-AzureFundamentals`) where this database will be stored.
  
* **Database Details**:
  * **Database name**: The unique identifier for your database instance (e.g., `sqldb-structured-data`).
  * **Server**: A logical container that manages security and credentials for one or more databases. If you don't have one, you click **Create new** to specify:
  * 🖥️ Creating an Azure SQL Server
    * **Server name**: A globally unique prefix ending in `.database.windows.net` (e.g., `apss`).
    * **Location**: The Azure region where the database server will be hosted (e.g., `Central India`).
    * **Authentication methods**: Configures how users and applications authenticate against the server:
      * **Use Microsoft Entra-only authentication**: Uses strictly Microsoft Entra ID (formerly Azure AD) credentials without maintaining traditional SQL logins. Recommended for high security.
      * **Use both SQL and Microsoft Entra authentication**: Allows access using either Microsoft Entra identities or traditional SQL server admin credentials.
      * **Use SQL authentication**: Relies entirely on traditional database administrator usernames and passwords.
    * **Set Microsoft Entra admin**: Selects an existing Microsoft Entra user, group, or application as the administrator.
    * **Server admin login**: The master SQL username for authentication (e.g., `Apoorva`).
    * **Password / Confirm password**: The secure administrator password.
  * **Want to use SQL elastic pool?**: Select **Yes** if you want to share compute and storage resources across multiple databases, or **No** if this database needs dedicated independent resources.

* Workload Environment
  * **Development**: Optimized for lower costs, testing, and non-production workloads.
  * **Production**: Sets robust defaults, high availability, and performance-tuned parameters for enterprise applications.

* Compute + Storage (Service Tiers)
  Clicking **Configure database** lets you choose your architectural tier and pricing model:
  * **General Purpose**: Balanced compute and storage option designed for standard, everyday workloads.
  * **Business Critical**: Designed for low-latency, high-availability transactional workloads with heavy read/write requirements.
  * **☁️ Azure SQL Database Hyperscale**: A highly scalable service tier that leverages cloud-scale architecture to decouple storage and compute, allowing storage to scale dynamically up to 100 TB+ without being constrained by local node sizes.

* Behavior when Free Offer Limit Reached
  When utilizing free monthly tier allowances (such as 100,000 vCore seconds and 32GB storage):
  * **Auto-pause the database until next month**: The database automatically pauses and becomes inaccessible once the free limit is hit, preventing any extra charges until the next calendar month.
  * **Continue using database for additional charges**: Keeps the database accessible, billing any overage usage at standard general-purpose serverless tier rates.

* Backup Storage Redundancy
  Controls how Point-In-Time Restore (PITR) and Long-Term Retention (LTR) backups are replicated for disaster recovery:
  * **Locally-redundant backup storage (LRS)**: Replicates your backups synchronously within a single physical data center.
  * **Zone-redundant backup storage (ZRS)**: Replicates backups across multiple availability zones within a region for higher availability.
  * **Geo-redundant backup storage (GRS)**: Replicates backups to a paired secondary region, enabling geo-restore capability during regional outages.
  * **Geo-Zone-redundant backup storage (GZRS)**: Combines zone redundancy with geo-replication for maximum enterprise-grade durability.

### 2. Networking Tab
The Networking tab determines how applications and users connect to your SQL database.

* **Connectivity method**:
  * **Public endpoint**: Allows access via the public internet (secured by firewalls and credentials).
  * **Private endpoint**: Connects your database securely inside a Virtual Network (VNet) using private IP addresses (recommended for enterprise production environments).
* **Firewall Rules**:
  * **Allow Azure services and resources to access this server**: Setting this to **Yes** enables other Azure services (like Azure Data Factory or Azure Functions) to communicate with your database.
  * **Add current client IP address**: Setting this to **Yes** automatically creates a firewall rule allowing your current IP address (e.g., your local machine or VS Code) to connect to the database server.

---

### 3. Security Tab
The Security tab configures advanced threat protection and data vulnerability management.

* **Microsoft Defender for SQL**:
  * Enables vulnerability assessments and advanced threat protection (detecting anomalous database access patterns, SQL injection attempts, etc.). Can be enabled or configured post-creation based on budget.
* **Identity**:
  * Configures Microsoft Entra ID (formerly Azure Active Directory) authentication alongside traditional SQL authentication for secure, centralized access control.

---

### 4. Additional Settings Tab
This tab handles data source initialization, collation rules, and data maintenance windows.

* **Use existing data**:
  * Allows you to provision your database from a blank database, a backup (sample data like `AdventureWorksLT`), or a point-in-time restore.
* **Collation**:
  * Defines the character set, sorting rules, and case-sensitivity for string data (e.g., `Latin1_General_CI_AS`).
* **Maintenance window**:
  * Specifies when system updates and patching can occur (e.g., System Default, Weekday, or Weekend windows).

---

### 5. Tags Tab
* **Definition**: Name/value pairs (metadata) assigned to resources to categorize them and streamline billing or resource management.
* **Example**: `Environment: Development`, `Project: DataEngineeringBootcamp`, or `Owner: DataTeam`.

---

### 6. Review + Create Tab
* **Validation**: Azure runs an automated validation check on your configurations.
* **Summary & Cost Estimate**: Displays estimated monthly costs and configuration summaries. Clicking **Create** provisions the database resources in your resource group.
