
# 01. Azure Fundamentals & Core Concepts

## Cloud: 

Center location to store the data. A vast global network of physical servers in secure data centers worldwide. Instead of storing data or running applications locally or on-premises, resources (compute, storage, network) are accessed remotely over the internet.

## Cloud Computing: 

The on-demand delivery of IT resources (compute power, storage, databases) over the internet with pay-as-you-go pricing, enabling dynamic scaling without maintaining physical data centers.

Cloud computing is the delivery of computing services—such as servers, storage, databases, networking, and analytics—over the internet. Instead of buying and maintaining your own hardware, you use a provider’s resources when you need them and pay according to your usage.

## 🔑 Key Characteristics of Cloud Computing
1. **On-demand self-service**: Users can provision computing resources (like virtual machines or databases) dynamically without requiring human intervention from the cloud provider.
2. **Broad network access**: Services are available over a network and accessed through standard mechanisms using heterogeneous client platforms (laptops, mobile devices, etc.).
3. **Resource pooling**: The provider’s computing resources are pooled to serve multiple consumers using a multi-tenant model, with different physical and virtual resources dynamically assigned and reassigned according to demand.
4. **Rapid elasticity**: Capabilities can be elastically provisioned and released to scale outward and inward rapidly based on demand.
5. **Measured service**: Cloud systems automatically control and optimize resource use by leveraging a metering capability at some level of abstraction appropriate to the type of service (e.g., storage, compute, bandwidth).

## 🌐 Cloud Deployment Models

* **Public Cloud**: Owned and operated by a third-party cloud service provider (e.g., Microsoft Azure, AWS), delivering computing resources like servers and storage over the internet.
* **Private Cloud**: Cloud infrastructure exclusively used by a single business or organization. It can be physically located at the organization's on-site data center or hosted by a third-party service provider.
* **Hybrid Cloud**: Combines public and private clouds, bound together by technology that allows data and applications to be shared between them.
* **Multi-Cloud**: Using multiple public cloud providers (e.g., combining Microsoft Azure and AWS) to avoid vendor lock-in and optimize workloads.

## 🛠️ Cloud Service Models

* **IaaS (Infrastructure as Service)**: 
  * *Definition*: Provides basic computing infrastructure—virtual machines (VMs), servers, storage, and networking.
  * *Management*: The cloud provider manages the physical hardware, data centers, and physical networking; you manage the OS, middleware, runtimes, and applications.
  * *Example*: Azure Virtual Machines.

* **PaaS (Platform as a Service)**: 
  * *Definition*: Provides a managed platform and environment allowing developers to build, test, and deploy applications without worrying about underlying infrastructure. **Primary model for Data Engineers**.
  * *Management*: The provider manages the hardware, OS, patching, and scaling; you focus purely on your application code and data configurations.
  * *Example*: Azure Data Factory, Azure Databricks, Azure SQL Database.

* **SaaS (Software as a Service)**: 
  * *Definition*: Delivers software applications over the internet, on-demand, typically via a subscription model.
  * *Management*: The vendor manages everything (servers, storage, networking, runtime, data, and application code); you simply use the software.
  * *Example*: Power BI, Microsoft 365.

## 🌍 Azure Physical Infrastructure & Availability

* **Region**: A set of datacenters deployed within a latency-defined perimeter and connected through a dedicated regional low-latency network.
* **Availability Zone (AZ)**: Physically separate locations within an Azure region made up of one or more datacenters equipped with independent power, cooling, and networking to ensure high availability.
* **Region Pair**: Each Azure region is paired with another region within the same geography (at least 300 miles away) to provide reliable replication and disaster recovery (DR).
* **Geography**: A discrete market, typically spanning two or more regions, that preserves data residency and compliance boundaries.

## 🗄️ Azure Storage Redundancy Options

* **LRS (Locally Redundant Storage)**: Replicates your data 3 times synchronously within a single datacenter in the primary region. Lowest cost, but least resilient against facility-wide outages.
* **ZRS (Zone-Redundant Storage)**: Replicates your data synchronously across 3 Azure availability zones in the primary region. Recommended for high availability (HA) for Data Lake storage (ADLS Gen2).
* **GRS (Geo-Redundant Storage)**: Copies your data synchronously 3 times in the primary region (via LRS), and then copies your data asynchronously 3 times to a secondary paired region.
* **GZRS (Geo-Zone-Redundant Storage)**: Combines ZRS (across 3 zones in the primary region) with GRS (asynchronous replication to the paired region). Ideal for maximum durability and enterprise disaster recovery.
